# Goal

When creating new scaling functions, or finetuning the parameters on an existing scaling function, it might help to look at the direct curve of how that function maps flux to [0,1]. You can do that with the `plot_curve` function, as explained below. These all assume units are in nanomaggies.

# Imports

```python
import sys
sys.path.append("..") #Point this to where the ensor.py file is located
from ensor import *
```

# Code

For example, the linear scale function looks like this. Note the x-axis is in magnitudes (i.e., logscale). 

```python
plot_curve(linear)
```

The asinh curve looks like this. You can see that it is linear for lower flux (high mag) values, and then behaves log-like at higher fluxes (low mag). 

```python
plot_curve(asinh)
```

We can also inspect the shape of the custom `root` function we defined in a previous tutorial:

```python
def root(x, n = 2):
    '''
    n is a tuneable parameter. If n == 2, then this is the same as sqrt. 
    '''
    return x**(1/n)
```

```python
plot_curve(root, n = 4)
```

You can compare different realizations of the log scale function like this: 

```python
plot_curve(log, a = 1, label = 'a=1', ax = plt.gca())
plot_curve(log, a = 10, label = 'a=10', ax = plt.gca())
plot_curve(log, a = 100, label = 'a=100',ax = plt.gca())

plt.legend()

plt.show()
```

You can compare different realizations of e.g., asinh, as follows: 

```python
plot_curve(asinh, Q = 20, stretch = 1, label = 'Q = 20; stretch = 1', ax = plt.gca())
plot_curve(asinh, Q = 80, stretch = 3, label = 'Q = 80; stretch = 3', ax = plt.gca())

plt.legend()

plt.show()
```

You can see the second version has a higher saturation limit, while fainter regions (below mag 24) are boosted a more.
